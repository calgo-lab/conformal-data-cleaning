```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from tab_err.error_mechanism import ECAR, ENAR, EAR

from dissertation_experiments.experiment import Experiment
from dissertation_experiments.data import split_columns_into_categorical_and_numerical
from dissertation_experiments.cleaner.forest import ConformalForestCleaner
from dissertation_experiments.config import DATA_PATH
```

```python
seed = 1

experiment = Experiment(name="test-seed", seed=seed, dataset_id=151)

train = experiment.X_train
test = experiment.X_test
perturbed, error_mask = experiment.perturb_and_get_perturbed_test_data_and_error_mask(error_rate=0.2, error_mechanism=ECAR)
```

```python
data_to_test = test

cleaner = ConformalForestCleaner(confidence_level=0.1).fit(train)
cleaned, cleaned_mask = cleaner.transform(data_to_test)

coverages = []
empty_set_fractions = []
average_set_sizes = []
relative_average_set_sizes = []

for column_name, prediction_sets in cleaner._prediction_sets.items():
    if column_name in experiment._categorical_columns:
        cardinality = len(experiment.X_train[column_name].unique())
        true_value_in_prediction_set = np.any(
            prediction_sets == data_to_test[column_name].to_numpy()[:, np.newaxis], axis=1
        )
        coverages.append(true_value_in_prediction_set.mean())

        average_set_sizes.append((~pd.DataFrame(prediction_sets).isna()).sum(axis=1).mean())
        relative_average_set_sizes.append(average_set_sizes[-1] / cardinality)
        empty_set_fractions.append(((~pd.DataFrame(prediction_sets).isna()).sum(axis=1) == 0).mean())

    elif column_name in experiment._numerical_columns:
        value_range = (
            experiment.X_train[column_name].max()
            - experiment.X_train[column_name].min()
        )
        true_value_in_prediction_range = (data_to_test[column_name].to_numpy() >= prediction_sets[:, 0]) & (
            data_to_test[column_name].to_numpy() <= prediction_sets[:, 1]
        )
        coverages.append(true_value_in_prediction_range.mean())
        average_set_sizes.append((prediction_sets[:, 1] - prediction_sets[:, 0]).mean())
        relative_average_set_sizes.append(average_set_sizes[-1] / value_range)
        empty_set_fractions.append(((prediction_sets[:, 1] - prediction_sets[:, 0]) == 0).mean())
```

```python
print(coverages)
print(empty_set_fractions)
print(average_set_sizes)
print(relative_average_set_sizes)
```
